```python
# remove warnings of data types
import pandas as pd
df6 = pd.read_csv("merged_dataset_movement_7.csv", low_memory=False)
```

```python
df_sample = df6.sample(n=100000, random_state=42)

X_s = df_sample.drop("MOVED", axis=1)
y_s = df_sample["MOVED"]

X_s_encoded = pd.get_dummies(X_s, drop_first=True)
```

```python
# fit the tree

from sklearn.tree import DecisionTreeClassifier

tree_default_sample = DecisionTreeClassifier(random_state=42)
tree_default_sample.fit(X_s_encoded, y_s)
```

```python
# Inspect

tree_default_sample.tree_.node_count, tree_default_sample.tree_.n_leaves
```

```python
# feature importances

import numpy as np

importances = tree_default_sample.feature_importances_
sorted_idx = np.argsort(importances)[::-1]

[(X_s_encoded.columns[i], importances[i]) for i in sorted_idx[:15]]
```

```python
# prunning the tree

path = tree_default_sample.cost_complexity_pruning_path(X_s_encoded, y_s)

ccp_alphas = path.ccp_alphas
impurities = path.impurities
```

```python
ccp_alphas = np.unique(ccp_alphas[ccp_alphas >= 0])
```

```python
# pick 25 evenly spaced alphas
alpha_grid = np.linspace(ccp_alphas.min(), ccp_alphas.max(), 25)
```

```python
# train pruned trees on the reduced grid

clfs = []

for ccp in alpha_grid:
    clf = DecisionTreeClassifier(random_state=42, ccp_alpha=ccp)
    clf.fit(X_s_encoded, y_s)
    clfs.append(clf)
```

```python
# Evaluate accuracy

from sklearn.model_selection import cross_val_score

alpha_scores = []

for clf in clfs:
    scores = cross_val_score(clf, X_s_encoded, y_s, cv=5)
    alpha_scores.append(scores.mean())
```

```python
best_alpha = alpha_grid[np.argmax(alpha_scores)]
best_alpha
```

```python
tree_pruned = DecisionTreeClassifier(random_state=42, ccp_alpha=best_alpha)
tree_pruned.fit(X_s_encoded, y_s)
```

```python
tree_pruned.tree_.node_count, tree_pruned.tree_.n_leaves
```

```python
from sklearn.model_selection import cross_val_score

scores_pruned = cross_val_score(tree_pruned, X_s_encoded, y_s, cv=5)
scores_pruned.mean()
```
