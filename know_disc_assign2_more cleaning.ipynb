```python
import pandas as pd
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.pipeline import Pipeline

# Load data
df = pd.read_csv("merged_dataset_movement_5.csv")

# Target
y = df["MOVED"]

# Features
X = df.drop(columns=["MOVED"])

# Identify categorical vs numeric
categorical_cols = X.select_dtypes(include=["object"]).columns
numeric_cols = X.select_dtypes(include=["int64", "float64"]).columns

# Preprocessing
preprocess = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse=False), categorical_cols),
        ("num", "passthrough", numeric_cols)
    ]
)


# Naive Bayes model
model = Pipeline(steps=[
    ("preprocess", preprocess),
    ("nb", GaussianNB())
])

# 10-fold CV
kf = KFold(n_splits=10, shuffle=True, random_state=42)

y_pred = cross_val_predict(model, X, y, cv=kf)

# Accuracy
accuracy = accuracy_score(y, y_pred)
print("10-fold CV Accuracy:", accuracy)

# Confusion matrix
cm = confusion_matrix(y, y_pred)
print("Confusion Matrix:")
print(cm)
```

```python
import pandas as pd
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.preprocessing import OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.pipeline import Pipeline

# Load data
df = pd.read_csv("merged_dataset_movement_5.csv")

# Drop high-cardinality columns
X = df.drop(columns=["MOVED", "COUNTYICP", "OCC"])

# Target
y = df["MOVED"]

# Identify categorical vs numeric
categorical_cols = X.select_dtypes(include=["object"]).columns
numeric_cols = X.select_dtypes(include=["int64", "float64"]).columns

# Preprocessing with OrdinalEncoder
preprocess = ColumnTransformer(
    transformers=[
        ("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), categorical_cols),
        ("num", "passthrough", numeric_cols)
    ]
)

# Naive Bayes model
model = Pipeline(steps=[
    ("preprocess", preprocess),
    ("nb", GaussianNB())
])

# 10-fold CV
kf = KFold(n_splits=10, shuffle=True, random_state=42)

y_pred = cross_val_predict(model, X, y, cv=kf)

# Accuracy
accuracy = accuracy_score(y, y_pred)
print("10-fold CV Accuracy:", accuracy)

# Confusion matrix
cm = confusion_matrix(y, y_pred)
print("Confusion Matrix:")
print(cm)
```

```python
# Dropping columns that are just creating noise and will not effect the model

drop_cols = [
    # Identifiers
    "household_id", "person_id", "hist_id", "hist_id_mother",
    "hist_id_father", "split_household_id", "link_1850",

    # High-cardinality geography
    "state", "city_name", "county_code", "county_nhg_code",

    # Relationship variables
    "relationship", "relationship_detail",

    # Occupation variables
    "occupation_code", "occupation_1950",

    # Group quarters details
    "gq_type", "gq_type_detail",

    # Birthplace detail
    "birthplace_detail",

    # Race detail
    "race_detail",

    # Household size variables
    "num_persons_recorded", "num_persons_household"
]

df_simplified = df.rename(columns=rename_map).drop(columns=drop_cols, errors="ignore")
```

```python
# Simplify sex into binary: male = 1, female = 0
df_simplified["sex_binary"] = df_simplified["sex"].apply(lambda x: 1 if x == 1 else 0)
```

```python
# simplifiying race categories 

def simplify_race(x):
    if x == 1:
        return "White"
    elif x == 2:
        return "Black"
    elif x == 3:
        return "Native"
    else:
        return "Other"

df_simplified["race_group"] = df_simplified["race"].apply(simplify_race)
```

```python
# US born or foreign born

def simplify_birthplace(x):
    if pd.isna(x):
        return "Unknown"
    elif x <= 56:
        return "US"
    else:
        return "Foreign"

df_simplified["birthplace_group"] = df_simplified["birthplace"].apply(simplify_birthplace)
```

```python
# Simplifying marriage status

def simplify_marital_status(x):
    if x == 1:
        return 1   # married
    else:
        return 0   # not married

df_simplified["marital_binary"] = df_simplified["marital_status"].apply(simplify_marital_status)
```

```python
def simplify_occupation(x):
    if pd.isna(x):
        return "Unknown"
    elif 0 <= x <= 99:
        return 1   # agricultural
    else:
        return 0   # non-agricultural

df_simplified["occupation_binary"] = df_simplified["occupation_code"].apply(simplify_occupation)
```

```python
# create age ranges

def age_range(x):
    if pd.isna(x):
        return "Unknown"
    elif x <= 14:
        return "0_14"
    elif x <= 24:
        return "15_24"
    elif x <= 44:
        return "25_44"
    elif x <= 64:
        return "45_64"
    else:
        return "65_plus"

df_simplified["age_group"] = df_simplified["age"].apply(age_range)
```
