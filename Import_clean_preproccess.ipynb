```python
# Disability counts by year and disability

import csv
from collections import defaultdict

# Initialize counters
disability_counts = {
    "1850": defaultdict(int),
    "1860": defaultdict(int),
    "1870": defaultdict(int),
    "1880": defaultdict(int),
}

with open("my_second_dataset.csv", "r", encoding="utf-8", errors="replace") as f:
    reader = csv.DictReader(f)

    for row in reader:
        yr = row["YEAR"]
        if yr not in disability_counts:
            continue

        # BLIND: only 2 = Yes
        if row["BLIND"] == "2":
            disability_counts[yr]["BLIND"] += 1

        # DEAF: 2 = Yes, 3 = Dumb only, 4 = Deaf only
        if row["DEAF"] in ("2", "3", "4"):
            disability_counts[yr]["DEAF"] += 1

        # IDIOTIC: only 2 = Yes
        if row["IDIOTIC"] == "2":
            disability_counts[yr]["IDIOTIC"] += 1

        # INSANE: only 2 = Yes
        if row["INSANE"] == "2":
            disability_counts[yr]["INSANE"] += 1

        # SICKNESS: only exists in 1880, only 2 = Yes
        if yr == "1880" and row["SICKNESS"] == "2":
            disability_counts[yr]["SICKNESS"] += 1

# Print results
for yr in ("1850", "1860", "1870", "1880"):
    print(f"\nYear {yr}:")
    for k, v in disability_counts[yr].items():
        print(f"  {k}: {v}")
```

```python
# total merged colums that will exsist within the merged data set

import csv

def get_columns(path):
    with open(path, "r", encoding="utf-8", errors="replace") as f:
        reader = csv.reader(f)
        return next(reader)

cols1 = get_columns("my_dataset.csv")
cols2 = get_columns("my_second_dataset.csv")

# Union of all columns
all_cols = list(dict.fromkeys(cols1 + cols2))

print("Total merged columns:", len(all_cols))
```

```python
# merging the two data sets

import csv

input_files = ["my_dataset.csv", "my_second_dataset.csv"]
output_file = "merged_dataset_1.csv"

with open(output_file, "w", newline="", encoding="utf-8") as out_f:
    writer = csv.DictWriter(out_f, fieldnames=all_cols)
    writer.writeheader()

    for file in input_files:
        with open(file, "r", encoding="utf-8", errors="replace") as in_f:
            reader = csv.DictReader(in_f)
            for row in reader:
                # Skip 1880
                if row["YEAR"] == "1880":
                    continue

                # Fill missing columns with empty strings
                full_row = {col: row.get(col, "") for col in all_cols}

                writer.writerow(full_row)

print("Merged file created:", output_file)
```

```python
# unique number of records in each data set, and combined number pre dedupe

import csv

def count_unique(path):
    seen = set()
    with open(path, "r", encoding="utf-8", errors="replace") as f:
        reader = csv.DictReader(f)
        for row in reader:
            yr = row["YEAR"]
            if yr in ("1850", "1860", "1870"):
                key = (row["YEAR"], row["SERIAL"], row["PERNUM"])
                seen.add(key)
    return seen

seen1 = count_unique("my_dataset.csv")
seen2 = count_unique("my_second_dataset.csv")

print("Dataset 1 unique (1850–1870):", len(seen1))
print("Dataset 2 unique (1850–1870):", len(seen2))

# Combined unique BEFORE removing duplicates
combined = seen1.union(seen2)
print("Combined unique (before dedupe):", len(combined))
```

```python
# dedupe data set
import csv

input_file = "merged_dataset_1.csv"
output_file = "merged_dataset_nodups_2.csv"

seen = set()

with open(input_file, "r", encoding="utf-8", errors="replace") as in_f, \
     open(output_file, "w", newline="", encoding="utf-8") as out_f:

    reader = csv.DictReader(in_f)
    writer = csv.DictWriter(out_f, fieldnames=reader.fieldnames)
    writer.writeheader()

    for row in reader:
        key = (row["YEAR"], row["SERIAL"], row["PERNUM"])

        if key in seen:
            continue  # skip duplicate

        seen.add(key)
        writer.writerow(row)

print("Final deduped file created:", output_file)
print("Total unique records:", len(seen))
```

```python
# making each disability category into single boolean 1 is disabled 0 is not

import csv

input_file = "merged_dataset_nodups_2.csv"
output_file = "merged_dataset_cleaned_3.csv"

with open(input_file, "r", encoding="utf-8", errors="replace") as in_f, \
     open(output_file, "w", newline="", encoding="utf-8") as out_f:

    reader = csv.DictReader(in_f)
    writer = csv.DictWriter(out_f, fieldnames=reader.fieldnames)
    writer.writeheader()

    for row in reader:

        # --- BLIND ---
        # 2 = Yes, everything else = No
        blind_val = row["BLIND"]
        row["BLIND"] = "1" if blind_val == "2" else "0"

        # --- DEAF ---
        # 2 = deaf & dumb, 3 = dumb only, 4 = deaf only
        deaf_val = row["DEAF"]
        row["DEAF"] = "1" if deaf_val in ("2", "3", "4") else "0"

        # --- IDIOTIC ---
        idiotic_val = row["IDIOTIC"]
        row["IDIOTIC"] = "1" if idiotic_val == "2" else "0"

        # --- INSANE ---
        insane_val = row["INSANE"]
        row["INSANE"] = "1" if insane_val == "2" else "0"

        # --- SICKNESS (only exists in 1880, but 1880 is removed) ---
        # If present, normalize anyway
        if "SICKNESS" in row:
            sickness_val = row["SICKNESS"]
            row["SICKNESS"] = "1" if sickness_val == "2" else "0"

        writer.writerow(row)

print("Cleaned dataset created:", output_file)
```

```python
import csv
from collections import defaultdict

input_file = "merged_dataset_cleaned_3.csv"
output_file = "county_disability_aggregates_4.csv"

# Structure:
# county_data[county][year][variable] = count
county_data = defaultdict(lambda: defaultdict(lambda: defaultdict(int)))

with open(input_file, "r", encoding="utf-8", errors="replace") as f:
    reader = csv.DictReader(f)

    for row in reader:
        yr = row["YEAR"]
        if yr not in ("1850", "1860", "1870"):
            continue

        county = row["COUNTYICP"]

        # Population count
        county_data[county][yr]["POP"] += 1

        # Disability counts (already normalized to 0/1)
        county_data[county][yr]["BLIND"] += int(row["BLIND"])
        county_data[county][yr]["DEAF"] += int(row["DEAF"])
        county_data[county][yr]["IDIOTIC"] += int(row["IDIOTIC"])
        county_data[county][yr]["INSANE"] += int(row["INSANE"])

# Write output
with open(output_file, "w", newline="", encoding="utf-8") as out_f:
    fieldnames = [
        "COUNTYICP", "YEAR", "POP",
        "BLIND", "DEAF", "IDIOTIC", "INSANE",
        "BLIND_RATE", "DEAF_RATE", "IDIOTIC_RATE", "INSANE_RATE"
    ]
    writer = csv.DictWriter(out_f, fieldnames=fieldnames)
    writer.writeheader()

    for county, years in county_data.items():
        for yr, stats in years.items():
            pop = stats["POP"]

            writer.writerow({
                "COUNTYICP": county,
                "YEAR": yr,
                "POP": pop,
                "BLIND": stats["BLIND"],
                "DEAF": stats["DEAF"],
                "IDIOTIC": stats["IDIOTIC"],
                "INSANE": stats["INSANE"],
                "BLIND_RATE": stats["BLIND"] / pop if pop else 0,
                "DEAF_RATE": stats["DEAF"] / pop if pop else 0,
                "IDIOTIC_RATE": stats["IDIOTIC"] / pop if pop else 0,
                "INSANE_RATE": stats["INSANE"] / pop if pop else 0,
            })

print("County-level disability file created:", output_file)
```

```python
import csv
from collections import defaultdict

input_file = "merged_dataset_cleaned_3.csv"

# Structure:
# county_disabilities[county] = { "BLIND": x, "DEAF": y, "IDIOTIC": z, "INSANE": w }
county_disabilities = defaultdict(lambda: defaultdict(int))

with open(input_file, "r", encoding="utf-8", errors="replace") as f:
    reader = csv.DictReader(f)

    for row in reader:
        yr = row["YEAR"]
        if yr not in ("1850", "1860", "1870"):
            continue

        county = row["COUNTYICP"]

        blind = int(row["BLIND"])
        deaf = int(row["DEAF"])
        idiotic = int(row["IDIOTIC"])
        insane = int(row["INSANE"])

        # Add counts by disability type
        county_disabilities[county]["BLIND"] += blind
        county_disabilities[county]["DEAF"] += deaf
        county_disabilities[county]["IDIOTIC"] += idiotic
        county_disabilities[county]["INSANE"] += insane

# Print only counties with at least one disabled person
for county, stats in county_disabilities.items():
    total_disabled = (
        stats["BLIND"] +
        stats["DEAF"] +
        stats["IDIOTIC"] +
        stats["INSANE"]
    )

    if total_disabled > 0:
        print(f"\nCounty {county}:")
        print(f"  BLIND: {stats['BLIND']}")
        print(f"  DEAF: {stats['DEAF']}")
        print(f"  IDIOTIC: {stats['IDIOTIC']}")
        print(f"  INSANE: {stats['INSANE']}")
```

```python
# Total linked individuals with a disability in ANY year

import csv
from collections import defaultdict

file = "merged_dataset_cleaned_3.csv"

# Track disability status by HIK
hik_disability = defaultdict(lambda: {"years": set(), "disabled_years": set()})

with open(file, "r", encoding="utf-8", errors="replace") as f:
    reader = csv.DictReader(f)
    for row in reader:
        hik = row["HIK"].strip()
        year = row["YEAR"]

        if hik == "":
            continue

        # Track years present
        hik_disability[hik]["years"].add(year)

        # Disability flags (already 0/1)
        blind = int(row["BLIND"])
        deaf = int(row["DEAF"])
        idiotic = int(row["IDIOTIC"])
        insane = int(row["INSANE"])

        if blind or deaf or idiotic or insane:
            hik_disability[hik]["disabled_years"].add(year)

# Count linked disabled individuals
linked_disabled = sum(1 for h, d in hik_disability.items() if len(d["disabled_years"]) > 0)

print("Total linked individuals with disability in ANY year:", linked_disabled)

```

```python
# total number of people that became disabled in any year (previoulsy not disabled then in subsequent year became disabled

became_disabled = 0
examples = []

for hik, info in hik_disability.items():
    years = sorted(info["years"])  # chronological order
    disabled_years = info["disabled_years"]

    if len(years) > 1:
        first_year = years[0]

        # If not disabled in first year but disabled later
        if first_year not in disabled_years and len(disabled_years) > 0:
            became_disabled += 1
            if len(examples) < 20:
                examples.append((hik, years, disabled_years))

print("Total individuals who became disabled:", became_disabled)

print("\nExamples of individuals who became disabled:")
for ex in examples:
    print(ex)
```

```python
# total number of individuals that moved in any year of the census

import csv
from collections import defaultdict

file = "merged_dataset_cleaned_3.csv"

# Store locations by HIK and YEAR
locations = defaultdict(lambda: {})

with open(file, "r", encoding="utf-8", errors="replace") as f:
    reader = csv.DictReader(f)
    for row in reader:
        hik = row["HIK"].strip()
        year = row["YEAR"]

        if hik == "":
            continue

        # Save location for this year
        locations[hik][year] = (
            row["STATEFIP"],
            row["COUNTYICP"]
        )

# Count movers
movers = 0

for hik, yrs in locations.items():
    if len(yrs) > 1:  # must appear in multiple years to be linked
        locs = list(yrs.values())
        if len(set(locs)) > 1:  # location changed
            movers += 1

print("Total linked individuals who moved:", movers)
```

```python
# new dataset of individuals who moved
# new column created for boolen didn't move = 0 did move = 1

import csv
from collections import defaultdict

input_file = "merged_dataset_cleaned_3.csv"
output_file = "merged_dataset_movement_5.csv"

# First pass: determine movement status for each HIK
locations = defaultdict(lambda: {})

with open(input_file, "r", encoding="utf-8", errors="replace") as f:
    reader = csv.DictReader(f)
    for row in reader:
        hik = row["HIK"].strip()
        year = row["YEAR"]
        if hik != "":
            locations[hik][year] = (
                row["STATEFIP"],
                row["COUNTYICP"]
            )

movement_flag = {}

for hik, yrs in locations.items():
    if len(yrs) > 1:
        locs = list(yrs.values())
        movement_flag[hik] = 1 if len(set(locs)) > 1 else 0
    else:
        movement_flag[hik] = 0  # only one year → cannot be a mover

# Second pass: write new dataset with MOVED column
with open(input_file, "r", encoding="utf-8", errors="replace") as f_in, \
     open(output_file, "w", encoding="utf-8", newline="") as f_out:

    reader = csv.DictReader(f_in)
    fieldnames = reader.fieldnames + ["MOVED"]
    writer = csv.DictWriter(f_out, fieldnames=fieldnames)
    writer.writeheader()

    for row in reader:
        hik = row["HIK"].strip()
        row["MOVED"] = movement_flag.get(hik, 0)
        writer.writerow(row)

print("New dataset created:", output_file)
```
