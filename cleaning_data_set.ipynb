```python
import pandas as pd
from collections import defaultdict


files = [
    "usa_00034.csv"
]

# Disability codes

blind_codes = {2}
deaf_codes = {2, 3, 4}
idiotic_codes = {2}
insane_codes = {2}

# storage dictonaries

year_counts = defaultdict(int)
year_disabled = defaultdict(int)
year_male = defaultdict(int)
year_female = defaultdict(int)
year_urban = defaultdict(int)
year_rural = defaultdict(int)
year_city = defaultdict(int)
year_noncity = defaultdict(int)
state_counts = defaultdict(lambda: defaultdict(int))

# Process each file

for file in files:
    print(f"Processing {file}...")

    for chunk in pd.read_csv(file, chunksize=100000, low_memory=False):
        # Ensure numeric
        for col in ['BLIND', 'DEAF', 'INSANE', 'SEX', 'URBAN', 'CITY', 'STATEFIP', 'YEAR']:
            if col in chunk.columns:
                chunk[col] = pd.to_numeric(chunk[col], errors='coerce').fillna(0).astype(int)
            else:
                chunk[col] = 0
        
        # disability flag
        disabled = (
            chunk['BLIND'].isin(blind_codes) |
            chunk['DEAF'].isin(deaf_codes) |
            chunk['IDIOTIC'].isin(idiotic_codes) |
            chunk['INSANE'].isin(insane_codes) 
        )

        # Iterate through rows
        for idx, row in chunk.iterrows():
            yr = row['YEAR']
            state = row['STATEFIP']

            year_counts[yr] += 1

            if disabled.loc[idx]:
                year_disabled[yr] += 1
            
            if row["SEX"] == 1:
                year_male[yr] += 1
            elif row["SEX"] == 2:
                year_female[yr] += 1

            if row["URBAN"] == 1:
                year_urban[yr] += 1
            else:
                year_rural[yr] += 1

            if row["CITY"] != 0:
                year_city[yr] += 1
            else:
                year_noncity[yr] += 1

            state_counts[yr][state] += 1

print("\n=== COMBINED SUMMARY TABLE ===")
print("YEAR | Total | Disabled | % Disabled | Male | Female | Urban | Rural | City | Non-city")
print("--------------------------------------------------------------------------")

for yr in sorted(year_counts):
    pct = year_disabled[yr] / year_counts[yr] * 100
    print(f"{yr} | {year_counts[yr]} | {year_disabled[yr]} | {pct:.4f}% | "
          f"{year_male[yr]} | {year_female[yr]} | "
          f"{year_urban[yr]} | {year_rural[yr]} | "
          f"{year_city[yr]} | {year_noncity[yr]}")

print("\n=== STATE COUNTS BY YEAR ===")
for yr in sorted(state_counts):
    print(f"\nYear {yr}:")
    for state, count in sorted(state_counts[yr].items()):
        print(f"  State {state}: {count}")
```

```python
import pandas as pd
from collections import defaultdict

file = "Linked_1850_1860.csv"
output = "Linked_1850_1860_with_migration.csv"

# Temporary storage: HISTID → list of rows
people = defaultdict(list)

# Read in chunks
for chunk in pd.read_csv(file, chunksize=100000, low_memory=False):
    for _, row in chunk.iterrows():
        histid = row["HISTID"]
        people[histid].append(row)

# Build final rows with migration indicator
final_rows = []

for histid, rows in people.items():
    if len(rows) == 2:  # linked across both years
        row_1850 = rows[0]
        row_1860 = rows[1]

        moved = int(row_1850["STATEFIP"] != row_1860["STATEFIP"])

        row_1850["moved_binary"] = moved
        row_1860["moved_binary"] = moved

        final_rows.extend([row_1850, row_1860])

    else:
        # Not linked across both years → cannot detect movement
        for r in rows:
            r["moved_binary"] = 0
            final_rows.append(r)

# Convert to DataFrame and save
df_out = pd.DataFrame(final_rows)
df_out.to_csv(output, index=False)

print("Finished:", output)
```

```python
# Keep only HISTIDs that appear in both years
linked = df.groupby("HISTID").filter(lambda g: g["YEAR"].nunique() == 2)

# Pivot STATEFIP by YEAR
state_pivot = linked.pivot(index="HISTID", columns="YEAR", values="STATEFIP")

# Check movement between 1860 and 1870
moved_state = (state_pivot[1860] != state_pivot[1870]).astype(int)

# Map back to full df
df["moved_binary"] = df["HISTID"].map(moved_state).fillna(0).astype(int)
```
